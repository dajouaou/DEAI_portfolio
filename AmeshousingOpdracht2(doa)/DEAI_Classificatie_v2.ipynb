Deze notebook bevat twee classificatiemodellen inclusief experimenten.

```
import pandas as pd  # Voor werken met dataframes
import numpy as np  # Voor numerieke berekeningen
from sklearn.model_selection import train_test_split  # Voor het splitsen van data
from sklearn.preprocessing import OneHotEncoder  # Voor omzetten categorische variabelen
from sklearn.compose import ColumnTransformer  # Om preprocessing te combineren
from sklearn.pipeline import Pipeline  # Om preprocessing + model te combineren
from sklearn.tree import DecisionTreeClassifier  # Het gekozen model
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score  # Evaluatiemetrics
```

```
df = pd.read_excel('AmesHousing.xlsx', sheet_name='AmesHousing')  # Lees het Excelbestand in als dataframe
df.head()  # Toon eerste 5 rijen om te controleren
```

```
features = ['OverallQual', 'GrLivArea', 'GarageCars', 'Neighborhood']  # Gekozen features (minstens 1 categorisch)
X = df[features]  # Selecteer feature kolommen
y_bin = df['GarageCars'] > 0  # Maak binaire target: wel of geen garage
y_multi = df['OverallQual']  # Multi-class target: kwaliteitsniveau
```

```
categorical_features = ['Neighborhood']  # Dit is de categorische feature
numeric_features = ['OverallQual', 'GrLivArea', 'GarageCars']  # Dit zijn numerieke features

preprocessor = ColumnTransformer(  # Combineer preprocessing
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features),  # One-hot encode categorische kolom
        ('num', 'passthrough', numeric_features)  # Laat numerieke kolommen onveranderd
    ])
```

```
X_train_bin, X_test_bin, y_train_bin, y_test_bin = train_test_split(  # Split voor binair model
    X, y_bin, test_size=0.2, random_state=42)  # 80% train, 20% test

X_train_multi, X_test_multi, y_train_multi, y_test_multi = train_test_split(  # Split voor multi model
    X, y_multi, test_size=0.2, random_state=42)
```

```
model = DecisionTreeClassifier(max_depth=5, min_samples_split=10, random_state=42)  # Decision tree met hyperparameters

pipeline = Pipeline(steps=[  # Combineer preprocessing en model
    ('preprocessing', preprocessor),  # Eerst preprocessing
    ('classifier', model)  # Daarna model
])
```

```
pipeline.fit(X_train_bin, y_train_bin)  # Train het model

y_pred_bin = pipeline.predict(X_test_bin)  # Maak voorspellingen

acc_bin = accuracy_score(y_test_bin, y_pred_bin)  # Bereken accuracy
prec_bin = precision_score(y_test_bin, y_pred_bin)  # Bereken precision
rec_bin = recall_score(y_test_bin, y_pred_bin)  # Bereken recall
f1_bin = f1_score(y_test_bin, y_pred_bin)  # Bereken F1-score

print(acc_bin, prec_bin, rec_bin, f1_bin)  # Print resultaten
```

```
pipeline.fit(X_train_multi, y_train_multi)  # Train multi-class model

y_pred_multi = pipeline.predict(X_test_multi)  # Maak voorspellingen

acc_multi = accuracy_score(y_test_multi, y_pred_multi)  # Accuracy
prec_multi = precision_score(y_test_multi, y_pred_multi, average='weighted')  # Weighted precision
rec_multi = recall_score(y_test_multi, y_pred_multi, average='weighted')  # Weighted recall
f1_multi = f1_score(y_test_multi, y_pred_multi, average='weighted')  # Weighted F1

print(acc_multi, prec_multi, rec_multi, f1_multi)  # Print resultaten
```

```
model_exp2 = DecisionTreeClassifier(max_depth=10, min_samples_split=5, random_state=42)  # Nieuwe hyperparameters

pipeline_exp2 = Pipeline(steps=[  # Nieuwe pipeline
    ('preprocessing', preprocessor),  # Zelfde preprocessing
    ('classifier', model_exp2)  # Nieuw model
])

pipeline_exp2.fit(X_train_bin, y_train_bin)  # Train opnieuw

y_pred_exp2 = pipeline_exp2.predict(X_test_bin)  # Nieuwe voorspellingen

print(accuracy_score(y_test_bin, y_pred_exp2))  # Toon nieuwe accuracy
```
